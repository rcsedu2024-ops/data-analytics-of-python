```python
# WRITE A PROGRAM TAKE THE INPUT FROM THE USER AND PRINT THE AREA,PERIMETER OF RECTANGLE,SQUARE,CIRCLE,TRIANGLE
length=float(input("Enter the length of the rectangle:"))
breadth=1/2
area=length*breadth
perimeter=2*(length+breadth)
print("perimeter is:",area)
print("perimeter is:",perimeter)
```

```python
side=float(input("Enter the side of the square: "))
area=side*side
perimeter=4*side
print("Area is:",area)
print("perimeter is:",perimeter)
```

```python
circle=float(input("Enter the radius of the circle: "))
area=3.14*circle*circle
perimeter=2*3.14*circle
print("Area is:",area)
print("perimeter is:",perimeter)
```

```python
tri=float(input("Enter the base of the triangle: "))
height=float(input("Enter the height of the triangle: "))
area=1/2*tri*height
perimeter=tri+height+((tri**2)+(height**2))**0.5
print("Area is:",area)
print("perimeter is:",perimeter)
```
